# 1- TF-IDF

<img src="https://drive.google.com/uc?export=view&id=1saSt0mMgOQ2ybbTms1lu_ruhUcBWkKzL" width="500">

TF-IDF stands for term frequency-inverse document frequency. It is a measure that discounts common words. Used in the fields of information retrieval (IR) and machine learning, that can quantify the importance of words in a document amongst a collection of documents (also known as a corpus).

### Components of TF-IDF

1. **TF (Term Frequency):**  
   Measures how frequently a term appears in a document.


$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$


2. **IDF (Inverse Document Frequency):**  
   Measures how important a term is across all documents. Words that appear in many documents get lower scores.

$$    \text{IDF}(t) = \log \left(\frac{\text{Number of all documents N}}{\text{Number of documents containing the term } t}\right) $$

# Cosine similarity

In NLP, Cosine similarity is a metric used to measure how similar the documents are.


$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|} $$

Where:  
$ A \cdot B $  = dot product of vectors A and B  
$ \|A\| $ = magnitude (length) of vector A  
$ \|B\| $ =  magnitude (length) of vector B

#### Intuition

- If vectors point in the **same direction**, cosine similarity = **1** (maximum similarity).
- If vectors are **orthogonal (90° apart)**, cosine similarity = **0** (no similarity).

### Task 1: Cosine Similarity
Use the Cosine Similarity method to determine how similar the following sentences are.

'This is the first document.',  
'This document is the second document.',  
'And this is the third one.',  
'Is this the first document?'  


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

sentences = ['This is the first document.',
             'This document is the second document.',
             'And this is the third one.',
             'Is this the first document?']

# Convert the sentences into TF-IDF vectors
task1_vectorizer = TfidfVectorizer()
task1_vectors = task1_vectorizer.fit_transform(sentences)

pd.DataFrame(task1_vectors.toarray(), columns=task1_vectorizer.get_feature_names_out(), index=sentences).round(3)
````

````python
# Calculate the cosine similarity between every pair of sentences
task1_similarity = cosine_similarity(task1_vectors)

task1_df = pd.DataFrame(task1_similarity, index=sentences, columns=sentences)
task1_df.round(3)
````

````python
# List the pairs from the most similar to the least similar
pairs = []
for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        pairs.append((task1_similarity[i, j], sentences[i], sentences[j]))

for score, first, second in sorted(pairs, reverse=True):
    print(f'{score:.3f}   "{first}"  <->  "{second}"')
````

**Result:**

- Sentence 1 (*"This is the first document."*) and sentence 4 (*"Is this the first document?"*) have a cosine similarity of **1.0**. They contain exactly the same words, only in a different order, and TF-IDF does not look at word order or punctuation, so both get the same vector.
- Sentence 2 is similar to sentences 1 and 4 (**0.647**) because they share `this`, `is`, `the` and `document`.
- Sentence 3 (*"And this is the third one."*) is the least similar to all the others (**0.225 – 0.308**), because it only shares the common words `this`, `is` and `the`, which have a low weight.

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


````python
documents = ['data science is one of the most important fields of science',
             'this is one of the best data science courses',
             'data scientists analyze data']

task2_vectorizer = TfidfVectorizer()
task2_vectors = task2_vectorizer.fit_transform(documents)

tfidf_df = pd.DataFrame(task2_vectors.toarray(),
                        columns=task2_vectorizer.get_feature_names_out(),
                        index=['doc_1', 'doc_2', 'doc_3'])
tfidf_df.round(3)
````

````python
# IDF of every word: words that appear in fewer documents get a higher IDF
idf_df = pd.DataFrame({'word': task2_vectorizer.get_feature_names_out(),
                       'documents containing the word': (task2_vectors.toarray() > 0).sum(axis=0),
                       'idf': task2_vectorizer.idf_})
idf_df.sort_values('idf', ascending=False).reset_index(drop=True).round(3)
````

````python
# The most important words of each document = the words with the highest TF-IDF score
for name, document in zip(tfidf_df.index, documents):
    scores = tfidf_df.loc[name]
    top_words = scores[scores > 0].sort_values(ascending=False).head(5)
    print(f'{name}: "{document}"')
    for word, score in top_words.items():
        print(f'   {word:<12} {score:.3f}')
    print()
````

**Result:**

- **doc_1:** the most important words are `science` and `of` (**0.488**), because each of them appears twice in the document. After them come `most`, `fields` and `important` (**0.321**), which appear only in this document.
- **doc_2:** the most important words are `best`, `courses` and `this` (**0.400**), because they appear only in this document.
- **doc_3:** the most important word is `data` (**0.641**), followed by `analyze` and `scientists` (**0.543**). `data` has the lowest IDF (**1.0**) because it appears in all three documents, but here it appears twice in a document of only four words, so its term frequency is very high. In doc_1 and doc_2, where it appears once, `data` has the lowest score (**0.190** and **0.236**).

So the TF-IDF score depends on two things together: a word is important when it is **frequent in the document** (high TF) and **rare in the other documents** (high IDF). Words like `data`, `is` and `the` that appear in most documents get a low IDF.

*Note:* scikit-learn uses a smoothed IDF, $\ln\left(\frac{1 + N}{1 + df}\right) + 1$, and normalizes each row, so the numbers are a little different from the simple formula at the top of the notebook.

````python
import re
import gensim
import pandas as pd
from nltk.tokenize import word_tokenize

simpsons = pd.read_csv('simpsons_script_lines.csv')

print('Shape of the dataset:', simpsons.shape)
print('Missing values:')
print(simpsons.isna().sum())
simpsons.head()
````

````python
# Some lines have no spoken words (they describe actions only), so they are removed
simpsons = simpsons.dropna(subset=['spoken_words']).reset_index(drop=True)
print('Number of lines with spoken words:', len(simpsons))
````

````python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

simpsons['clean_words'] = simpsons['spoken_words'].apply(clean_text)
simpsons[['spoken_words', 'clean_words']].head(10)
````

````python
# Tokenize every line into words
simpsons_tokens = []

for line in simpsons['clean_words']:
    simpsons_tokens.append(word_tokenize(line))

print('Number of tokenized lines:', len(simpsons_tokens))
print('Total number of tokens   :', sum(len(t) for t in simpsons_tokens))
print(simpsons_tokens[0])
````

````python
tokens = simpsons_tokens

Skip_gram_model = gensim.models.Word2Vec(tokens, min_count = 1, vector_size = 100, window = 5, sg = 1)

print(Skip_gram_model)
print('Vocabulary size:', len(Skip_gram_model.wv))
print("Shape of the vector of 'homer':", Skip_gram_model.wv['homer'].shape)
````

# What I Learned from Lab 5

In this lab I learned how to **represent text as numbers** so that a machine learning model can use it. I worked with two techniques, **TF-IDF** and **Word2Vec**, and I used **cosine similarity** to compare the vectors.

### TF-IDF
- TF-IDF gives every word in a document a score made of two parts: **TF** (how often the word appears in the document) and **IDF** (how rare the word is across all the documents).
- A word gets a high score when it is frequent in one document and rare in the others. Words that appear everywhere get a low score.
- `TfidfVectorizer().fit_transform()` returns a sparse matrix with one row per document and one column per word. `get_feature_names_out()` gives the words of the columns.
- scikit-learn uses a smoothed IDF and normalizes every row, so the numbers are not exactly the same as the simple formula, but the idea is the same.

### Cosine Similarity
- Cosine similarity measures the angle between two vectors: **1** means the same direction (very similar) and **0** means nothing in common.
- **Task 1:** *"This is the first document."* and *"Is this the first document?"* got a similarity of **1.0**, because they contain the same words and TF-IDF ignores word order and punctuation. The second sentence was close to them (**0.647**) and the third sentence was the least similar (**0.225 – 0.308**).
- This showed me a weakness of TF-IDF: a statement and a question with the same words look identical to it.

### Task 2: Important Words with TF-IDF
- In the first document the most important words were `science` and `of` (**0.488**), because each appears twice.
- In the second document they were `best`, `courses` and `this` (**0.400**), which appear only in that document.
- In the third document `data` was the highest (**0.641**) because it is half of a four-word document, even though it has the lowest IDF (**1.0**) since it appears in all three documents. In the other two documents `data` had the lowest score.
- So the score depends on both parts: a high TF can still win against a low IDF in a very short document.

### Word2Vec
- Word2Vec is a small neural network that learns a dense vector for every word from the words around it. It has two architectures: **CBOW** and **Skip-gram** (`sg=1`).
- Unlike TF-IDF, words that are used in similar contexts get similar vectors, so the vectors capture **meaning**.
- The parameters: `vector_size` is the length of the vector, `window` is how many words around the target are used, and `min_count` ignores words that appear fewer times than this number.

**Conclusion:** TF-IDF is simple and shows which words are important in a document, but it treats every word as unrelated to the others. Word2Vec learns the meaning of words from their context, so similar words get close vectors. The right choice depends on the task and on how much text is available.
